```python
### 27. Find the largest integer less than or equal to the square root of a given number.
import math
num = int(input("Enter a number: "))
result = math.isqrt(num)
print(f"The largest integer less than or equal to the square root of {num} is {result}")
large = 0
for i in range(int(result)):
  if large <= i:
    large = i
print(f"The largest integer less than or equal to the square root of {num} is {large}")
```

```python
### 28. Calculate compound interest using math.pow().
amount = int(input("Enter the amount :"))
interest = int(input("Enter the interest :"))
time = int(input("Enter the time :"))
compound_interest = amount * (pow((1 + interest / 100), time))
print(f"Compound interest is {compound_interest}")
```

```python
### 29. Find the area and circumference of a circle using math.pi.
radius = int(input("enter the radius :"))
area = math.pi * math.pow(radius,2)
print(f"Area of a circle is {area}:")
circumference = 2 * math.pi * radius
print(f"Circumference of a circle is {circumference}")
```
