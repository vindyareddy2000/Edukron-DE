```python
### Question 15. Using `filter()`, `map()`, and `reduce()`:
### 1. Extract students scoring more than 70.
### 2. Extract only their marks.
### 3. Calculate the average marks.
students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]
marks = list(map(lambda x : x['marks'], list(filter(lambda x : x['marks'] > 70 , students))))
total = reduce(lambda x , y : x + y , marks, 0)
average = total / len(marks)
print(average)
```

```python
### Question 16. Using `filter()`, `map()`, and `reduce()`:
### 1. Extract products currently in stock.
### 2. Calculate each product's inventory value using `price × stock`.
### 3. Calculate the total inventory value.
products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]
in_stock = list(filter(lambda x : x['stock'] > 0, products))
inventory = list(map(lambda x : x['price'] * x['stock'], in_stock))
reduce(lambda x , y : x + y, inventory, 0)
```

```python
### Question 17. Using `filter()`, `map()`, and `reduce()`:
### 1. Extract products belonging to the Electronics category.
### 2. Apply a 20% discount.
### 3. Calculate the total discounted value.
products = [
    {"name": "Laptop", "price": 55000, "category": "Electronics"},
    {"name": "Chair", "price": 4500, "category": "Furniture"},
    {"name": "Monitor", "price": 18000, "category": "Electronics"},
    {"name": "Desk", "price": 8500, "category": "Furniture"},
    {"name": "SSD", "price": 6500, "category": "Electronics"}
]
electronics = list(filter(lambda x : x['category'] == 'Electronics', products))
after_discount = list(map(lambda x : x['price'] * 20 / 100, electronics))
reduce(lambda x , y : x + y, after_discount, 0)
```

```python
### Question 18. Using `filter()` and `reduce()`:
### 1. Extract delivered orders.
### 2. Calculate the total delivered revenue.
orders = [
    {"order_id": 1001, "amount": 1200, "status": "Delivered"},
    {"order_id": 1002, "amount": 3500, "status": "Pending"},
    {"order_id": 1003, "amount": 8500, "status": "Delivered"},
    {"order_id": 1004, "amount": 4200, "status": "Cancelled"},
    {"order_id": 1005, "amount": 7600, "status": "Delivered"},
    {"order_id": 1006, "amount": 5100, "status": "Pending"}
]
delivered = list(filter(lambda x : x['status'] == 'Delivered', orders))
reduce(lambda x , y : x + y['amount'], delivered, 0)
```

```python
### Question 19. Using `filter()`, `map()`, and `reduce()`:
### 1. Extract transactions whose amount is greater than 5000.
### 2. Add 18% GST to every selected transaction.
### 3. Calculate the grand total after GST.
transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]
amount = list(filter(lambda x : x['amount'] > 5000, transactions))
with_gst = list(map(lambda x : x['amount'] + x['amount'] * 18 / 100, amount))
reduce(lambda x , y : x + y, with_gst, 0)
```

```python
### Question 20. Using `filter()`, `map()`, and `reduce()`:
### 1. Extract active salespeople from the South region.
### 2. Calculate a 5% incentive for each selected salesperson.
### 3. Create a list containing salesperson name, sales amount, and incentive.
### 4. Calculate the total incentive payable.
sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]
active = list(filter(lambda x : x['region'] == 'South' and x['active'] == True, sales_records))
incentive = list(map(lambda x : {'salesperson' : x['salesperson'], 'sales' : x['sales'], 'incentive' : x['sales'] * 5 / 100}, active))
reduce(lambda x , y : x + y['incentive'], incentive, 0)
```
